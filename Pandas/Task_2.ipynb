```python
import pandas as pd
```

```python
Student_Data = pd.read_csv("Student.csv")
```

```python
Student_Data
```

```python
Student_Data["Class"] = Student_Data["roll_no"].str.split("/").str[1]
```

```python
Student_Data["Obtain"] = Student_Data.iloc[:,2:-1].sum(axis = 1)
```

```python
Student_Data
```

```python
Student_Data["Total"] = 100
```

```python
Student_Data
```

```python
Student_Data["Percentage"] = (Student_Data.Obtain/Student_Data.Total)*100
```

```python
Student_Data
```

```python
def Grade(x):
    if x >= 80:
        return "A"
    elif x >= 60:
        return "B"
    elif x >= 40:
        return "C"
    elif x < 40:
        return "F"
        
```

```python
Student_Data["Grades"] = Student_Data.Percentage.apply(Grade)
```

```python
Student_Data["Rank"] = Student_Data.Percentage.sort_values().index+1
Student_Data
```

```python
grp_std = Student_Data.groupby("Class")
```

```python
DSAI = grp_std.get_group("DSAI")
```
